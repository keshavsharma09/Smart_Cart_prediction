```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("smartcart_customers.csv")

df.head()
```

```python
df.isnull().sum()
```

# Data Preprocessing

## 1. Handle Missiing values

```python
df["Income"] = df["Income"].fillna(df["Income"].median())
```

```python
df.head()
```

# Feature Engineering

```python
# Age
df["Age"] = 2026-df["Year_Birth"]
```

```python
# Customer Joining Date

df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"],dayfirst=True)

refrence_date = df["Dt_Customer"].max()

df["Customer_Tenure_Daye"] = (refrence_date - df["Dt_Customer"]).dt.days
```

```python
# spending 

df["Spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
```

```python
df["Total_Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
# Education

df["Education"].value_counts()

df["Education"] = df["Education"].replace({
    "Basic": "Undergraduate","2n Cycle": "Undergraduate",
    "Graduation": "Graduate",
    "PhD": "Postgraduate","Master": "Postgraduate"
})
```

```python
# Marital Status

df["Marital_Status"].value_counts()

df["Living_with"] = df["Marital_Status"].replace({
    "Married": "Partner","Together": "Partner",
    "Single": "Alone","Divorced":"Alone",
    "Widow": "Alone","Alone": "Alone","Absurd" :"Alone","YOLO":"Alone"
})
```

### Drop Columns

```python
df.columns
```

```python
col = ["ID", "Year_Birth", "Marital_Status","Kidhome","Teenhome","Dt_Customer"]
spending_cols = ["MntWines","MntFruits","MntMeatProducts","MntFishProducts","MntSweetProducts","MntGoldProds"]

cols_to_drop = col + spending_cols

df_cleaned = df.drop(columns=cols_to_drop)
```

```python
df_cleaned.shape
```

```python
df_cleaned.head()
```

### Outliers|

```python
cols =  ["Income","Recency","Response","Age","Spending","Total_Children"]

# Relatives plots or some features- pair plots

sns.pairplot(df_cleaned[cols])
```

```python
# Removeoutliers

print("data size with outliers:",len(df_cleaned))

df_cleaned = df_cleaned[ (df_cleaned["Age"] < 90) ]
df_cleaned = df_cleaned[ (df_cleaned["Income"] < 600_000) ]

print("data size without outliers:",len(df_cleaned))
```

## Encoding

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()

cat_cols = ["Education", "Living_with"]

enc_cols = ohe.fit_transform(df_cleaned[cat_cols])
```

```python
enc_df = pd.DataFrame(enc_cols.toarray(), columns=ohe.get_feature_names_out(cat_cols), index=df_cleaned.index)
```

```python
df_encoded = pd.concat([df_cleaned.drop(columns=cat_cols), enc_df], axis=1)
```

```python
df_encoded.shape
```

```python
df_encoded.head()
```

## Scaling

```python
from sklearn.preprocessing import StandardScaler
```

```python
X = df_encoded
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

# Visualization


```python
X_scaled.shape
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=3)

X_pca = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_ratio_
```

```python
# plot

fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")
ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2])

ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_xlabel("3d projection")
```

## Analyze K value

## 1. Elbow Method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
    
```

```python
knee = KneeLocator(range(1, 11), wcss, curve="convex", direction="decreasing")
optimal_k = knee.elbow
```

```python
print("best k =", optimal_k)
```

```python
# plot 

plt.plot(range(1,11), wcss, marker="o")
plt.xlabel("K")
plt.ylabel("Wcss")
```

## 2. siluatte Score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)
    score = silhouette_score(X_pca, label)
    scores.append(score)

# plot
plt.plot(range(2,11), scores, marker="o")
plt.xlabel("K")
plt.ylabel("Silhouette_score")
```

```python
# combined plot

k_range = range(2, 11)

fig, ax1 = plt.subplots(figsize=(8, 6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("wcss")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker="x", color="red", linestyle="--")
ax2.set_ylabel("SS")
```

## Clustring

```python
# K_means
kmeans = KMeans(n_clusters=4, random_state=42)
labels_means = kmeans.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2], c=labels_means)
```

```python
# Agglomerative Clustring
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8, 6))

ax = fig.add_subplot(111, projection="3d")

ax.scatter(X_pca[:,0],X_pca[:,1],X_pca[:,2], c=labels_agg)
```
